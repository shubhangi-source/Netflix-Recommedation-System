# Part 1: Data Cleaning and Preparation
This notebook covers data loading, initial inspection, text cleaning, handling missing values, and date parsing.

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

# Load the dataset from CSV

```python

df = pd.read_csv("../data/data/netflix_titles.csv")
```

# Statistical Summary

# Standardize column names

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
print(df.head(4))
```

# check missing values

```python
print(df.isnull().sum().sort_values(ascending=False))
```

# Duplicate Records

```python
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
num_cols=df.select_dtypes(include=np.number).columns.tolist()
cat_cols=df.select_dtypes(include='object').columns.tolist()
print(num_cols)
print(cat_cols)
```

# Clean text columns

```python

text_cols = ["type", "title", "director", "cast", "country", "rating", "listed_in"]
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()
```

# Fill missing categorical values 

```python
df["director"]=df["director"].fillna("Unknown")
df["cast"]=df["cast"].fillna("Not Available")

mode_country=df["country"].mode()[0]
df["country"]=df["country"].fillna(mode_country)

mode_rating=df["rating"].mode()[0]
df["rating"]=df["rating"].fillna(mode_country)
```

# Convert date_added to datetime

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

# fill missing values in numerical columns

```python
median_date=df["date_added"].dropna().median()
df["date_added"]=df["date_added"].fillna(median_date)
```

```python
mode_duration = df["duration"].mode()[0]

df["duration"] = df["duration"].fillna(mode_duration)
```

# Check Missing Values Again

```python
df.isnull().sum()
```

```python
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month_name()
```

# Saved netflix cleaned csv

```python
df.to_csv("netflix_cleaned.csv", index=False)
print("Saved netflix_cleaned.csv")
```
